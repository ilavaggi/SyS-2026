Generamos y graficamos x1 y x2

```python
import numpy as np

def x1(t):
    x1t = np.array(len(t))
    x1t = np.where(t<0, 0, t)
    x1t = np.where(x1t>3, 0, x1t)
    return x1t
```

```python
def x2(t):
    x2t = np.array(len(t))
    x2t = np.where(t<-1,0,t)
    x2t = np.where(t>=-1,1,x2t)
    x2t = np.where(t>=0,3,x2t)
    x2t = np.where(t>=1,2,x2t)
    x2t = np.where(t>=2,1,x2t)
    x2t = np.where(t>3,0,x2t)

    return x2t
```

```python
import matplotlib.pyplot as plt
plt.style.use("ggplot")


coso,eje = plt.subplots(1,2,figsize=(20,6))

t = np.linspace(-8,8,7000)
x1t = x1(t)
x2t = x2(t)

eje[0].plot(t,x1t,markevery=0.5)
eje[1].plot(t,x2t)

eje[0].set_title("X1")
eje[1].set_title("X2")

eje[0].set_xticks(range(-8, 9)) # para que me ponga las rayitas cada 1 en lugar de cada 2
eje[1].set_xticks(range(-8, 9))

plt.show()
```
