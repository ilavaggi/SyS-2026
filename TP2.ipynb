### a)

```python
import numpy as np

def a(n):
    return (0+1j)**n / n
```

```python
import matplotlib.pyplot as plt
n = np.array(range(1,50))
an = a(n)

plt.figure(figsize=(20,6))
plt.style.use("ggplot")
plt.stem(n,an)
plt.yticks(np.arange(-1.0,1.1,0.1))
plt.show()

# converge a 0
```

```python
def b(n):
    return (1+1j)**n / n
```

```python
n = np.array(range(1,50))
bn = b(n)

plt.figure(figsize=(20,6))
plt.style.use("ggplot")
plt.stem(n,bn)
#plt.yticks(np.arange(-1.0,1.1,0.1))
plt.show()
```
