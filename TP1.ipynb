Generamos y graficamos x1 y x2

```python
import numpy as np

def x1(t):
    x1t = np.array(len(t))
    x1t = np.where(t<0, 0, t)
    x1t = np.where(x1t>3, 0, x1t)
    return x1t

```

```python
def x2(t):
    x2t = np.array(len(t))
    x2t = np.where(t<-1,0,t)
    x2t = np.where(t>=-1,1,x2t)
    x2t = np.where(t>=0,3,x2t)
    x2t = np.where(t>=1,2,x2t)
    x2t = np.where(t>=2,1,x2t)
    x2t = np.where(t>3,0,x2t)

    return x2t
```

```python
import matplotlib.pyplot as plt
plt.style.use("ggplot")
plt.figure(figsize=(16.0,4.0))

t = np.linspace(-8,8,7000)
x1t = x1(t)
plt.plot(t,x1t)
```

```python
x1t = x1(t-2)
plt.plot(t,x1t)
```

```python
x1t = x1(2*t)
plt.plot(t,x1t)
```

```python
x1t = x1(t/2)
plt.plot(t,x1t)
```

```python
x1t = x1(-t)
plt.plot(t,x1t)
```

```python
x1t = x1(-t+2)
plt.plot(t,x1t)
```

```python
y = x1(t)+x2(t)
plt.plot(t,y)
```

```python
y = x1(t)*x2(t)
plt.plot(t,y)
```
