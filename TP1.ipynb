## 1 ------------------------------------------------------------------
Generamos y graficamos x1 y x2

```python
import numpy as np

def x1(t):
    x1t = np.array(len(t))
    x1t = np.where(t<0, 0, t)
    x1t = np.where(x1t>3, 0, x1t)
    return x1t
```

```python
def x2(t):
    x2t = np.array(len(t))
    x2t = np.where(t<-1,0,t)
    x2t = np.where(t>=-1,1,x2t)
    x2t = np.where(t>=0,3,x2t)
    x2t = np.where(t>=1,2,x2t)
    x2t = np.where(t>=2,1,x2t)
    x2t = np.where(t>3,0,x2t)

    return x2t
```

```python
import matplotlib.pyplot as plt
plt.style.use("ggplot")


coso,eje = plt.subplots(1,2,figsize=(20,6))

t = np.linspace(-8,8,7000)
x1t = x1(t)
x2t = x2(t)

eje[0].plot(t,x1t,markevery=0.5)
eje[1].plot(t,x2t)

eje[0].set_title("X1")
eje[1].set_title("X2")

eje[0].set_xticks(range(-8, 9)) # para que me ponga las rayitas cada 1 en lugar de cada 2
eje[1].set_xticks(range(-8, 9))

plt.show()
```

## 2 ------------------------------------------------------------------
armamos las funciones base

```python
# ni hace falta usar numpy, son poquitos valores. hacemo' una lista
# edit: no puedo hacer las cuentas sobre el range, asi que tengo que convertirlo si o si a array :(
# pero convierto el n antes de pasarlo a la funcion. Ver mas adelante

def x1n(n):
    base = [1,2,3,0,0,2,2]
    yn=[]
    for i in n:
        if 0 <= i < len(base):
            yn.append(base[i])
        else:
            yn.append(0)
    return np.array(yn)
```

```python
# vamos con la otra
def x2n(n):
    base = [-2,-2,2,2,0,-2,0]
    yn=[]
    for i in n:
        if -2 <= i < len(base)-2:
            yn.append(base[i+2]) # tengo que sumarle al indice, por que el dibujito estyá "corrido" del cero en 2 espacios
        else:
            yn.append(0)
    return np.array(yn)

# Si a alguien se le ocurre una manera mas comoda de hacerlo, jelp

```

```python
n = range(-5,8)

# le cambio los colorcitos, solo por que si
plt.style.use("bmh")

coso2,eje = plt.subplots(1,2,figsize=(20,6))
eje[0].stem(n,x1n(n))
eje[1].stem(n,x2n(n))

eje[0].set_title("X1")
eje[1].set_title("X2")

eje[0].set_xticks(n)
eje[1].set_xticks(n)


plt.show()
```
